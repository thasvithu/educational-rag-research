# Understanding Embeddings from Scratch

```python
# ============================================================
# CELL 1: THE CORE PROBLEM — WHY DO WE NEED EMBEDDING?
# ============================================================

# Computers cannot understand human language directly.
# Computers only understand numbers.
#
# Think about this:
# You ask a question: "What are the scholarship requirements?"
# The computer cannot compare this text to chunks in the database.
# The computer needs NUMBERS to do comparison.
#
# So we need a way to convert text into numbers.
# That process is called EMBEDDING.
#
# EMBEDDING = Converting text into a list of numbers
#             where the numbers CAPTURE THE MEANING of the text
#
# The key word is MEANING.
# Not just any random numbers.
# Numbers that represent what the text MEANS.

print("Computers understand numbers, not words.")
print("Embedding converts words into numbers.")
print("Similar words get similar numbers.")
print("That is the entire concept of embedding.")
```

```python
# ============================================================
# CELL 2: SIMPLE DEMONSTRATION — BEFORE USING ANY MODEL
# ============================================================

# Let us first understand WHY simple approaches do not work.
# Then you will appreciate WHY we need a model.

# NAIVE APPROACH 1: Assign a number to each word
# -----------------------------------------------
simple_dictionary = {
    "scholarship": 1,
    "student":     2,
    "university":  3,
    "examination": 4,
    "requirements":5,
}

sentence = "scholarship requirements"
naive_representation = [simple_dictionary.get(word, 0)
                         for word in sentence.split()]

print("NAIVE APPROACH — Assign numbers to words:")
print(f"Sentence: '{sentence}'")
print(f"Naive numbers: {naive_representation}")
print()
print("PROBLEM with this approach:")
print("'scholarship rules' = [1, 0]")
print("'scholarship requirements' = [1, 5]")
print("These look different but mean almost the same thing!")
print("A naive dictionary cannot capture MEANING.")
print()
```

```python
# NAIVE APPROACH 2: Count word frequency (Bag of Words)
# -------------------------------------------------------
from collections import Counter

sentence1 = "scholarship covers full tuition fees"
sentence2 = "tuition fees scholarship full covers"

count1 = Counter(sentence1.split())
count2 = Counter(sentence2.split())

print("NAIVE APPROACH 2 — Count words (Bag of Words):")
print(f"Sentence 1: '{sentence1}'")
print(f"Sentence 2: '{sentence2}'")
print(f"Word counts are identical: {count1 == count2}")
print()
print("PROBLEM with this approach:")
print("Word order is lost completely.")
print("'Dog bites man' and 'Man bites dog'")
print("would get the SAME representation — clearly wrong!")
```

```python
# ============================================================
# CELL 3: WHAT A REAL EMBEDDING LOOKS LIKE
# ============================================================

# A real embedding is a list of floating point numbers
# called a VECTOR.
# Each number captures a different dimension of meaning.
#
# For example — the model all-MiniLM-L6-v2 produces
# a vector of 384 numbers for any text input.
#
# These 384 numbers together represent the full meaning
# of the input text.

import numpy as np

# This is what a REAL embedding looks like
# (these specific numbers are just for illustration)
example_embedding = [
    0.234, -0.156,  0.891, -0.023,  0.445,
   -0.312,  0.678,  0.123, -0.789,  0.234,
    # ... 374 more numbers ...
]

print("A real embedding vector looks like this:")
print(f"[{example_embedding[0]}, {example_embedding[1]}, "
      f"{example_embedding[2]}, {example_embedding[3]}, "
      f"{example_embedding[4]}, ... 379 more numbers]")
print()
print("Total numbers in one embedding: 384")
print("Each number is between -1 and +1 approximately")
print("Together these 384 numbers represent the full MEANING")
print("of the input text")
```

```python
# ============================================================
# CELL 4: WHY DO WE NEED A MODEL TO GENERATE EMBEDDINGS?
# ============================================================

# This is the most important question.
# Why can't we just assign numbers manually?
#
# The answer: Understanding meaning requires learning from
# enormous amounts of text.
#
# The all-MiniLM-L6-v2 model was trained on:
# - Hundreds of millions of sentence pairs
# - It learned which sentences mean similar things
# - It learned which words appear in similar contexts
# - This learning is stored in the model's WEIGHTS
#   (millions of internal numbers)
#
# When you give text to the model:
# The model uses its learned weights to process the text
# and output a 384-number vector that captures meaning.
#
# WITHOUT the model — you cannot capture meaning.
# You can only count words or assign arbitrary numbers.
# Those approaches do not work for intelligent search.

print("WHY WE NEED A MODEL:")
print()
print("Step 1: Someone trained a model on millions of")
print("        sentence pairs like:")
print("        'scholarship covers tuition'")
print("        'the award pays for education fees'")
print("        → Model learns these mean the same thing")
print()
print("Step 2: The model stores this learning in")
print("        millions of internal numbers called WEIGHTS")
print()
print("Step 3: When you give the model new text")
print("        it uses those weights to produce a vector")
print("        that correctly represents meaning")
print()
print("Step 4: You get 384 numbers that capture meaning")
print("        NO HUMAN could assign these manually")
print("        It took millions of examples to learn them")
```

```python
# ============================================================
# CELL 9: WHAT HAPPENS INTERNALLY — STEP BY STEP
# ============================================================

# This cell explains the internal process
# so your team understands what .encode() is actually doing

from transformers import AutoTokenizer
import torch

tokenizer = AutoTokenizer.from_pretrained(
    'sentence-transformers/all-MiniLM-L6-v2'
)

sentence = "scholarship covers tuition fees"

print("INTERNAL PROCESS — What happens inside .encode()")
print("=" * 60)
print()
```

```python
# STEP 1: Tokenization
print("STEP 1: TOKENIZATION")
print("  Split text into tokens (pieces the model understands)")
tokens = tokenizer.tokenize(sentence)
token_ids = tokenizer.encode(sentence)
print(f"  Input text: '{sentence}'")
print(f"  Tokens: {tokens}")
print(f"  Token IDs (numbers): {token_ids}")
print()
print("  Notice: Words are split into subword pieces")
print("  'scholarship' might become ['scholar', '##ship']")
print("  This helps with unknown words")
print()
```

```python
# STEP 2: Through transformer layers
print("STEP 2: THROUGH 6 TRANSFORMER LAYERS")
print("  Each layer processes the tokens")
print("  Each layer adds understanding of:")
print("  - Word relationships")
print("  - Context")
print("  - Meaning")
print("  This happens 6 times (L6 = 6 layers)")
print()
```

```python
# STEP 3: Pooling
print("STEP 3: MEAN POOLING")
print("  After 6 layers — each token has its own vector")
print(f"  Input had {len(tokens)} tokens")
print(f"  Each token now has a 384-number vector")
print(f"  Total: {len(tokens)} × 384 numbers")
print()
print("  Mean pooling takes the AVERAGE across all tokens")
print(f"  Result: ONE vector of 384 numbers")
print("  This represents the WHOLE sentence meaning")
print()
```

```python
# STEP 4: Normalization
print("STEP 4: NORMALIZATION")
print("  The 384 numbers are scaled to have length = 1")
print("  This is called L2 normalization")
print("  It makes similarity comparison more accurate")
print()
print("  FINAL OUTPUT: 384 normalized numbers")
print("  = The embedding of your sentence")
```

```python
# ============================================================
# CELL 10: PROVE THAT SIMILAR TEXTS GET SIMILAR NUMBERS
# ============================================================

import numpy as np
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')

# Three pairs of sentences
# Pair 1: Very similar meaning
sentence_A = "Students must have a GPA of 3.5 for the scholarship."
sentence_B = "The minimum grade point average for scholarship eligibility is 3.5."

# Pair 2: Somewhat related
sentence_C = "Examination rules must be followed by all students."

# Pair 3: Completely different
sentence_D = "The canteen is open from 8am to 5pm."

# Generate embeddings for all
emb_A = model.encode(sentence_A)
emb_B = model.encode(sentence_B)
emb_C = model.encode(sentence_C)
emb_D = model.encode(sentence_D)

# Calculate cosine similarity
# Cosine similarity measures how similar two vectors are
# Score of 1.0 = identical meaning
# Score of 0.0 = completely unrelated

def cosine_similarity(vec1, vec2):
    """
    Cosine similarity between two vectors.
    Returns a score between -1 and 1.
    Higher = more similar meaning.
    """
    dot_product = np.dot(vec1, vec2)
    magnitude1 = np.linalg.norm(vec1)
    magnitude2 = np.linalg.norm(vec2)
    return dot_product / (magnitude1 * magnitude2)

sim_AB = cosine_similarity(emb_A, emb_B)
sim_AC = cosine_similarity(emb_A, emb_C)
sim_AD = cosine_similarity(emb_A, emb_D)
```

```python
print("PROVING THAT SIMILAR TEXTS GET SIMILAR NUMBERS")
print("=" * 60)
```

```python
print(f"Sentence A: '{sentence_A}'")
print(f"Sentence B: '{sentence_B}'")
print(f"Sentence C: '{sentence_C}'")
print(f"Sentence D: '{sentence_D}'")
```

```python
print(f"Similarity A vs B (very similar meaning): {sim_AB:.4f}")
print(f"Similarity A vs C (related topic):        {sim_AC:.4f}")
print(f"Similarity A vs D (completely different): {sim_AD:.4f}")
```

```python
# ============================================================
# CELL 11: EMBED A FULL UNIVERSITY DOCUMENT CHUNK
# ============================================================

# This is exactly what happens in our RAG research
# Each chunk from a university handbook gets embedded

chunks_from_handbook = [
    "Students who wish to apply for the Bachelor of ICT degree must meet the following requirements. The applicant must have completed Advanced Level examination with passes in three subjects.",
    "The university offers merit-based scholarships to outstanding students. To be eligible for the scholarship, a student must maintain a Grade Point Average of 3.5 or above throughout the academic year.",
    "All students must follow the examination rules strictly. Students must arrive at the examination hall at least fifteen minutes before the scheduled start time.",
]

print("EMBEDDING UNIVERSITY HANDBOOK CHUNKS")
print("=" * 60)
print()
print(f"Number of chunks to embed: {len(chunks_from_handbook)}")
print()

# Embed all chunks at once
# batch processing is faster than one by one
chunk_embeddings = model.encode(chunks_from_handbook)

print(f"Shape of all embeddings: {chunk_embeddings.shape}")
print(f"  {len(chunks_from_handbook)} chunks × 384 dimensions each")
print()

for i, (chunk, embedding) in enumerate(
        zip(chunks_from_handbook, chunk_embeddings)):
    print(f"Chunk {i+1}:")
    print(f"  Text: '{chunk[:60]}...'")
    print(f"  Embedding shape: {embedding.shape}")
    print(f"  First 5 numbers: {embedding[:5]}")
    print()

print("These embeddings are now ready to be stored in FAISS.")
print("When a student asks a question,")
print("the question is also embedded,")
print("and FAISS finds the most similar chunk.")
```
