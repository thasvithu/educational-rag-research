# Structure-Aware Chunking

```python
from langchain.text_splitter import MarkdownHeaderTextSplitter
from langchain.text_splitter import RecursiveCharacterTextSplitter
```

```python

# CELL 3: What is Structure-Aware Chunking?
# ------------------------------------------------------------
# Instead of splitting by size — this method respects the
# STRUCTURE of the document.
#
# It splits at natural boundaries like:
# - Chapter headings
# - Section headings
# - Paragraph breaks
#
# For university handbooks — each section becomes one chunk.
# "Admission Requirements" → one chunk
# "Scholarship Eligibility" → one chunk
# "Examination Rules" → one chunk
#
# ADVANTAGE: Each chunk is a complete, coherent topic
# DISADVANTAGE: Chunk sizes vary — some very long, some short
# ------------------------------------------------------------
```

```python
markdown_text = """
# Chapter 1: Admission Requirements

Students who wish to apply for the Bachelor of Information and 
Communication Technology degree must meet the following requirements. 
The applicant must have completed Advanced Level examination with 
passes in three subjects. Mathematics must be one of the three 
subjects passed. The minimum score required is 200 marks combined 
across all three subjects.

# Chapter 2: Scholarship Eligibility

The university offers merit-based scholarships to outstanding students. 
To be eligible for the scholarship, a student must maintain a Grade 
Point Average of 3.5 or above throughout the academic year. The 
scholarship covers full tuition fees for one academic year. It also 
includes a monthly allowance of five thousand rupees. Students who 
receive disciplinary warnings are not eligible to apply.

# Chapter 3: Examination Rules

All students must follow the examination rules strictly. Students 
must arrive at the examination hall at least fifteen minutes before 
the scheduled start time. Mobile phones and electronic devices are 
strictly prohibited inside the examination hall. Students caught 
cheating will face immediate disqualification. The examination 
results will be published within four weeks after the examination date.
"""
```

```python
# Create structure-aware chunker
headers_to_split_on = [
    ("#", "Chapter"),      # Split at # headings
    ("##", "Section"),     # Split at ## headings
]

chunker = MarkdownHeaderTextSplitter(
    headers_to_split_on=headers_to_split_on
)

chunks = chunker.split_text(markdown_text)
```

```python
# Show results
print("=" * 60)
print("STRUCTURE-AWARE CHUNKING RESULTS")
print("=" * 60)
print(f"Total chunks created: {len(chunks)}")
print()

for i, chunk in enumerate(chunks):
    print(f"--- CHUNK {i+1} ---")
    print(f"Metadata: {chunk.metadata}")
    print(f"Content:\n{chunk.page_content}")
    print(f"[Length: {len(chunk.page_content)} characters]")
    print()
```

```python
# Compare with fixed-size
print("=" * 60)
print("WHY THIS IS BETTER THAN FIXED-SIZE")
print("=" * 60)
print("""
Fixed-size: Cuts every 200 characters — ignores chapter boundaries
Structure-aware: Cuts at chapter headings — each chunk = one complete topic

For university handbooks this is very natural because:
- Admission Requirements = complete idea = one chunk
- Scholarship Eligibility = complete idea = one chunk  
- Examination Rules = complete idea = one chunk

When a student asks "what are the exam rules?"
The retriever finds the Examination Rules chunk
And the answer is COMPLETE — not cut in half
""")
```
