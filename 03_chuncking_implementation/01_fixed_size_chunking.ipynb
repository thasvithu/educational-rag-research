```python
# Import library
from langchain.text_splitter import CharacterTextSplitter
```

```python
# Our sample educational text
sample_text = """
Chapter 1: Admission Requirements

Students who wish to apply for the Bachelor of Information and 
Communication Technology degree must meet the following requirements. 
The applicant must have completed Advanced Level examination with 
passes in three subjects. Mathematics must be one of the three 
subjects passed. The minimum score required is 200 marks combined 
across all three subjects.

Chapter 2: Scholarship Eligibility

The university offers merit-based scholarships to outstanding students. 
To be eligible for the scholarship, a student must maintain a Grade 
Point Average of 3.5 or above throughout the academic year. The 
scholarship covers full tuition fees for one academic year. It also 
includes a monthly allowance of five thousand rupees. Students who 
receive disciplinary warnings are not eligible to apply.

Chapter 3: Examination Rules

All students must follow the examination rules strictly. Students 
must arrive at the examination hall at least fifteen minutes before 
the scheduled start time. Mobile phones and electronic devices are 
strictly prohibited inside the examination hall. Students caught 
cheating will face immediate disqualification. The examination 
results will be published within four weeks after the examination date.
"""
```

```python
# What is Fixed-Size Chunking?
# ------------------------------------------------------------
# Fixed-size chunking splits the document into chunks of 
# exactly the same size — measured in characters or tokens.
# It does NOT care about sentences, paragraphs, or topics.
# It just counts and cuts.
#
# ADVANTAGE: Very simple and fast
# DISADVANTAGE: Can cut in the middle of a sentence or topic
# ------------------------------------------------------------
```

```python
# Create the fixed-size chunker
# chunk_size = how many characters per chunk
# chunk_overlap = 0 means no overlap between chunks
chunker = CharacterTextSplitter(
    chunk_size=200,       # Each chunk = 200 characters
    chunk_overlap=0,      # No overlap
    separator=""          # No specific separator — just count
)
```

```python
# Apply chunking to our sample text
chunks = chunker.split_text(sample_text)
```

```python
# Show results
print("=" * 60)
print("FIXED-SIZE CHUNKING RESULTS")
print("=" * 60)
print(f"Total chunks created: {len(chunks)}")
print()

for i, chunk in enumerate(chunks):
    print(f"--- CHUNK {i+1} ---")
    print(chunk)
    print(f"[Length: {len(chunk)} characters]")
    print()
```

```python
# Show the PROBLEM with fixed-size chunking
print("=" * 60)
print("THE PROBLEM — OBSERVE CAREFULLY")
print("=" * 60)
print("""
Look at the chunks above carefully.
You will notice that some chunks:
- Start in the middle of a sentence
- End in the middle of a sentence
- Mix content from two different chapters

This is called BOUNDARY FRAGMENTATION.
This is one of the failure modes in our research.

Example: "It also includes a monthly allowance"
— The word "It" refers to the scholarship
— But if this appears in a separate chunk
— The AI does not know what "It" refers to
— This leads to wrong or incomplete answers
""")
```
