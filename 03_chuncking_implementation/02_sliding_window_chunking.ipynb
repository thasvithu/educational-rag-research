# Sliding Window Chunking

```python
from langchain.text_splitter import CharacterTextSplitter
```

```python
sample_text = """
Chapter 1: Admission Requirements

Students who wish to apply for the Bachelor of Information and 
Communication Technology degree must meet the following requirements. 
The applicant must have completed Advanced Level examination with 
passes in three subjects. Mathematics must be one of the three 
subjects passed. The minimum score required is 200 marks combined 
across all three subjects.

Chapter 2: Scholarship Eligibility

The university offers merit-based scholarships to outstanding students. 
To be eligible for the scholarship, a student must maintain a Grade 
Point Average of 3.5 or above throughout the academic year. The 
scholarship covers full tuition fees for one academic year. It also 
includes a monthly allowance of five thousand rupees. Students who 
receive disciplinary warnings are not eligible to apply.

Chapter 3: Examination Rules

All students must follow the examination rules strictly. Students 
must arrive at the examination hall at least fifteen minutes before 
the scheduled start time. Mobile phones and electronic devices are 
strictly prohibited inside the examination hall. Students caught 
cheating will face immediate disqualification. The examination 
results will be published within four weeks after the examination date.
"""
```

```python
# What is Sliding Window Chunking?
# ------------------------------------------------------------
# Sliding window is similar to fixed-size chunking BUT
# it adds OVERLAP between consecutive chunks.
#
# Example with chunk_size=200 and chunk_overlap=50:
# Chunk 1: characters 0   to 200
# Chunk 2: characters 150 to 350   (goes back 50 characters)
# Chunk 3: characters 300 to 500   (goes back 50 characters)
#
# The overlap ensures that if important content is near
# a boundary — it appears in BOTH neighboring chunks.
#
# ADVANTAGE: Reduces boundary fragmentation
# DISADVANTAGE: Creates duplicate content — larger storage
# ------------------------------------------------------------
```

```python
# Compare three overlap settings
overlap_settings = [0, 50, 100]

for overlap in overlap_settings:
    print("=" * 60)
    print(f"chunk_size=200, chunk_overlap={overlap}")
    print("=" * 60)

    chunker = CharacterTextSplitter(
        chunk_size=200,
        chunk_overlap=overlap,
        separator=""
    )

    chunks = chunker.split_text(sample_text)
    print(f"Total chunks: {len(chunks)}")

    # Show first two chunks to see overlap effect
    for i, chunk in enumerate(chunks[:2]):
        print(f"\n--- CHUNK {i+1} ---")
        print(chunk)
    print()
```

```python
# Show the overlap clearly
print("=" * 60)
print("SEEING THE OVERLAP CLEARLY")
print("=" * 60)

chunker = CharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=50,
    separator=""
)
chunks = chunker.split_text(sample_text)

print("Last 50 characters of Chunk 1:")
print(repr(chunks[0][-50:]))
print()
print("First 50 characters of Chunk 2:")
print(repr(chunks[1][:50]))
print()
print("These should look SIMILAR — that is the overlap working.")
```

```python
# The advantage and remaining problem
print("""
ADVANTAGE OF SLIDING WINDOW:
If important content is near a boundary — it appears in both 
neighboring chunks. So retrieval is less likely to miss it.

REMAINING PROBLEM:
Even with overlap — the chunks still do not respect topic 
boundaries. Chapter content can still be mixed together.
This is why we need smarter chunking methods.
""")
```
